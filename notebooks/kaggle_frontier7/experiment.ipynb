Este notebook reproduce partidas oficiales en la nube contra el V48 sin modificar como control emparejado y contra rivales públicos, exige ausencia de errores, latencia inferior a 1000 ms y ninguna regresión por rival, y solo entonces escribe `submission.tar.gz`.
```python
from pathlib import Path
import os,json,base64,lzma
os.chdir("/kaggle/working")
payload=json.loads(lzma.decompress(base64.b64decode('')))
for name,data in payload.items():
    p=Path(name)
    p.parent.mkdir(parents=True,exist_ok=True)
    p.write_bytes(data.encode("utf-8"))
print("extracted",len(payload),"files")
```
